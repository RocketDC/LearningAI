```python
# Simple demonstration of adding noise to an image tensor
import torch
import matplotlib.pyplot as plt

image = torch.rand(3, 64, 64) # Dummy image tensor
noise = torch.randn_like(image)

# Add varying levels of noise
fig, axs = plt.subplots(1, 4, figsize=(12, 3))
for i, beta in enumerate([0.1, 0.3, 0.6, 0.9]):
    noisy_image = torch.sqrt(1 - torch.tensor(beta)) * image + torch.sqrt(torch.tensor(beta)) * noise
    axs[i].imshow(noisy_image.permute(1, 2, 0).numpy().clip(0, 1))
    axs[i].set_title(f"Noise beta={beta}")
    axs[i].axis('off')
plt.show()
```
